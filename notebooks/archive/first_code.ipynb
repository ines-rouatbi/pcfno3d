```python
# ============================================================
# Importation des bibliothèques nécessaires
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import glob
import warnings

warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.4f}'.format)

print('✅ Bibliothèques chargées avec succès')
```

```python
# ============================================================
# Définition des chemins vers les données
# DATA_FOLDER : dossier contenant les 30 fichiers journaliers
# GRID_FILE   : fichier unique de la grille spatiale
# ============================================================

DATA_FOLDER = r"C:\Users\inesr\OneDrive\Desktop\PFE\aneo_physical_db_multitime_v3_progress\aneo_physical_db_multitime_v3_progress\data_4x"

GRID_FILE = os.path.join(DATA_FOLDER, "aneo_grid_points.csv")

# Lister tous les fichiers journaliers (30 jours)
daily_files = sorted(glob.glob(os.path.join(DATA_FOLDER, "aneo_observations_day_*.csv")))

print(f'✅ Dossier trouvé')
print(f'   Fichiers journaliers détectés : {len(daily_files)}')
print()
for f in daily_files:
    print(f'   📄 {os.path.basename(f)}')
```

```python
# ============================================================
# Chargement du fichier de grille spatiale (unique)
# 500 000 points — relief, rugosité, type de terrain
# ============================================================

df_grid = pd.read_csv(GRID_FILE)

print(f'✅ Grille spatiale chargée')
print(f'   Dimensions : {df_grid.shape[0]:,} lignes × {df_grid.shape[1]} colonnes')
print()
df_grid.head(5)
```

```python
# ============================================================
# Chargement d'un seul fichier journalier pour vérification
# Avant de charger les 30 jours, on vérifie la structure
# ============================================================

df_sample = pd.read_csv(daily_files[0])

print(f'✅ Fichier test chargé : {os.path.basename(daily_files[0])}')
print(f'   Dimensions : {df_sample.shape[0]:,} lignes × {df_sample.shape[1]} colonnes')
print()
print('=== TIMESTAMPS ===')
print(df_sample['timestamp_utc'].unique())
print()
print('=== HEURES UTC ===')
print(sorted(df_sample['hour_utc'].unique()))
print()
df_sample.head(3)
```

```python
# ============================================================
# Chargement et concaténation des 30 fichiers journaliers
# Résultat attendu : 30 × 2 000 000 = 60 000 000 lignes
# ============================================================

print('⏳ Chargement des 30 fichiers en cours...')
print()

all_days = []

for i, filepath in enumerate(daily_files):
    day_name = os.path.basename(filepath)
    df_day = pd.read_csv(filepath)
    all_days.append(df_day)
    print(f'   [{i+1:02d}/30] ✅ {day_name} — {df_day.shape[0]:,} lignes')

df_obs = pd.concat(all_days, ignore_index=True)

print()
print(f'✅ Chargement terminé')
print(f'   Dimensions totales : {df_obs.shape[0]:,} lignes × {df_obs.shape[1]} colonnes')
print(f'   Mémoire estimée    : {df_obs.memory_usage(deep=True).sum() / 1e9:.2f} GB')
```

```python
# ============================================================
# Vérification de la cohérence globale après concaténation
# Nombre de jours, timestamps, points par jour
# ============================================================

print('=== JOURS COUVERTS ===')
print(f'day_index min : {df_obs["day_index"].min()} | max : {df_obs["day_index"].max()}')
print(f'Jours uniques : {df_obs["day_index"].nunique()}')
print()

print('=== TIMESTAMPS UNIQUES ===')
print(df_obs['timestamp_utc'].unique()[:8], '...')
print()

print('=== RÉPARTITION PAR HEURE ===')
print(df_obs.groupby('hour_utc')['point_id'].count())
print()

print('=== SOURCE TYPE ===')
print(df_obs['source_type'].value_counts())
```

```python
# ============================================================
# Jointure des observations avec la grille spatiale
# via le point_id — ajoute x, y, z_terrain, rugosité, terrain
# ============================================================

df_merged = df_obs.merge(
    df_grid[['point_id', 'x_m', 'y_m', 'z_terrain_m',
             'z_rel_m', 'z0_m', 'terrain_type', 'land_use_code',
             'slope_deg', 'aspect_deg']],
    on='point_id',
    how='left'
)

# Position 3D réelle de chaque mesure
df_merged['z_abs_m'] = df_merged['z_terrain_m'] + 25

print(f'✅ Jointure effectuée')
print(f'   Dimensions après jointure : {df_merged.shape[0]:,} × {df_merged.shape[1]}')
print()
print('=== VÉRIFICATION z_abs_m ===')
print(f'Min : {df_merged["z_abs_m"].min():.2f} m | Max : {df_merged["z_abs_m"].max():.2f} m')
print()
df_merged[['point_id', 'x_m', 'y_m', 'z_terrain_m', 'z_abs_m', 'terrain_type']].head(5)
```
